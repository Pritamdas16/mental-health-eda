# **Project Summary -**

Mental health is one of the most critical yet underrepresented topics in the modern tech industry. Despite growing awareness, many employees still hesitate to seek professional help due to stigma, lack of workplace support, or fear of negative career consequences. This project performs a comprehensive Exploratory Data Analysis (EDA) on the OSMI Mental Health in Tech Survey dataset collected in 2014, which captures responses from 1,259 tech industry professionals across multiple countries.
The primary goal of this analysis is to understand the patterns and factors associated with mental health treatment-seeking behavior among tech workers, and to evaluate how well employers are currently supporting their employees' mental health needs.
The dataset contains 27 features covering a wide range of variables including respondent demographics (age, gender, country), employment details (company size, remote work, tech company), mental health history (family history, work interference), employer support systems (benefits, wellness programs, anonymity, leave policies), and social perceptions (willingness to discuss with coworkers or supervisors, fear of consequences).
The analysis begins with thorough data cleaning. The Age column contained severe outliers — values as low as -1726 and as high as 100 billion — which were removed by keeping only realistic values between 15 and 75. The Gender column had over 40 unique entries due to free-text input, which were standardized into three clean categories: Male, Female, and Other/Non-binary. Missing values in work_interfere and self_employed were filled appropriately.
Following the UBM (Univariate, Bivariate, Multivariate) analysis framework, 20+ meaningful visualizations were created. Univariate analysis revealed that the dataset is male-dominated (79.1%), most respondents are between 25 and 40 years old, and the treatment variable is nearly balanced (50.5% Yes, 49.5% No) — making it ideal for future machine learning classification tasks. The United States accounts for 60% of all respondents.
Bivariate analysis uncovered several important relationships. Females seek treatment at a significantly higher rate (62.8%) compared to males (47.4%). Family history of mental illness emerged as the strongest single predictor — 76.2% of those with a family history sought treatment, compared to only 30.7% of those without. Work interference also showed a strong correlation: employees who reported that mental health "Often" interferes with their work had an 87.8% treatment rate.
Analysis of workplace support revealed a concerning trend — a large proportion of employees are unaware of the mental health benefits their employer provides. Wellness programs are rarely discussed, and anonymity protection is largely unknown. This suggests that even where support exists, poor communication prevents employees from utilizing it.
The stigma analysis showed that approximately 35% of respondents fear negative professional consequences from discussing mental health with their employer, compared to only 19% for physical health. This significant gap highlights that mental health stigma remains a real and persistent barrier in the tech industry, even in 2014.
Overall, this EDA provides actionable insights for tech organizations: proactively communicate mental health benefits, create stigma-free environments, and pay special attention to employees showing signs of work interference. The balanced target variable also makes this dataset a strong candidate for a future classification project predicting treatment-seeking behavior.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# 1. Age: remove outliers
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df = df[(df['Age'] >= 15) & (df['Age'] <= 75)].copy()

# 2. Gender: standardize
def clean_gender(g):
    g = str(g).strip().lower()
    male_terms = ['male', 'm', 'man', 'cis male', 'make', 'mal', 'maile',
                  'msle', 'malr', 'mail', 'cis man', 'male (cis)', 'male-ish']
    female_terms = ['female', 'f', 'woman', 'cis female', 'cis-female/femme',
                    'female (cis)', 'femail', 'femake', 'female (trans)',
                    'trans woman', 'trans-female', 'women']
    if any(g == t for t in male_terms) or g.startswith('male'):
        return 'Male'
    elif any(g == t for t in female_terms) or g.startswith('female') or g == 'woman':
        return 'Female'
    else:
        return 'Other/Non-binary'

df['Gender_clean'] = df['Gender'].apply(clean_gender)

# 3. Fill missing values
df['work_interfere'] = df['work_interfere'].fillna('Unknown')
df['self_employed'] = df['self_employed'].fillna('No')

# 4. Drop comments column (95% missing, not useful for EDA)
df.drop(columns=['comments'], inplace=True)

print("Data Wrangling Complete!")
print("Final Shape:", df.shape)
print("\nGender distribution:")
print(df['Gender_clean'].value_counts())
````

### What all manipulations have you done and insights you found?

The following manipulations were performed to make the dataset analysis-ready. First, the Age column contained extreme outliers including negative values and values in the billions — all records outside the range of 15 to 75 were removed, bringing the row count from 1,259 to 1,251. Second, the Gender column had over 40 inconsistent free-text entries such as "m", "Male", "maile", "femail", "Trans woman" etc. — these were standardized into three clean categories: Male (989), Female (251), and Other/Non-binary (11). Third, missing values in work_interfere (264 missing) were filled with "Unknown" to preserve those rows for analysis, and self_employed (18 missing) was filled with "No" as the default assumption. Fourth, the comments column was dropped entirely as 1,095 out of 1,259 values were missing, making it unsuitable for structured analysis. After wrangling, the dataset is clean, consistent, and ready for Univariate, Bivariate, and Multivariate analysis.

Based on the EDA findings, the following actionable suggestions are recommended to help tech organizations achieve their business objective of improving employee mental health and reducing stigma:
1. Proactively Communicate Mental Health Benefits

4. Train Managers and Supervisors
Employees are more willing to discuss mental health with coworkers than with supervisors, which indicates a trust gap in management. Organizations should invest in mental health awareness training for all managers so employees feel safe discussing struggles with their direct supervisors.
5. Strengthen Anonymity and Confidentiality Policies
35% of employees fear negative professional consequences from disclosing mental health issues. Companies should clearly communicate and legally strengthen their confidentiality policies around mental health disclosures, ensuring employees that seeking help will never negatively impact their career.
6. Build a Data-Driven Mental Health Risk Model
The EDA reveals that treatment-seeking behavior can be predicted using a combination of variables — family history, work interference, gender, and workplace support. Organizations should invest in building an internal ML-based early warning system to proactively identify at-risk employees and intervene before conditions worsen.
7. Improve Wellness Programs
Most respondents reported that their employer has never discussed mental health as part of a wellness program. Regular mental health workshops, mindfulness sessions, Employee Assistance Programs (EAPs), and mental health days should be introduced as standard practice.
